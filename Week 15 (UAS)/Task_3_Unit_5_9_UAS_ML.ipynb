## **Unit 6: The Tokenizers Library**

# **Tujuan:**

Mempelajari cara melakukan tokenisasi pada data teks menggunakan pustaka Tokenizers dari Hugging Face.

# **Penjelasan:**

Tokenisasi adalah proses memecah teks menjadi unit-unit kecil yang disebut token. Dalam unit ini, kita akan menggunakan tokenizer berbasis WordPiece.

Pengecekan File data.txt: Sebelum memulai pelatihan tokenizer, kode memeriksa apakah file data.txt sudah ada. Jika tidak, kode akan membuat file tersebut dengan beberapa kalimat sampel.

Inisialisasi Tokenizer: Kita membuat tokenizer menggunakan model WordPiece.

Pre-tokenizer: Kita menggunakan pre-tokenizer berbasis spasi untuk memisahkan kata-kata.

Melatih Tokenizer: Tokenizer dilatih menggunakan file teks eksternal.

Mengkodekan Teks: Kita mengkodekan teks sampel untuk melihat hasil tokenisasi.

```python
from tokenizers import Tokenizer
from tokenizers.models import WordPiece
from tokenizers.trainers import WordPieceTrainer
from tokenizers.pre_tokenizers import Whitespace
import os

# Membuat file data.txt jika belum ada
if not os.path.exists("data.txt"):
    with open("data.txt", "w") as f:
        f.write("Hello, how are you?\n")
        f.write("I am learning about tokenizers from Hugging Face.\n")
        f.write("Tokenization is an important step in NLP.\n")

# Inisialisasi tokenizer
tokenizer = Tokenizer(WordPiece())

# Mengatur pre-tokenizer
tokenizer.pre_tokenizer = Whitespace()

# Melatih tokenizer
trainer = WordPieceTrainer(vocab_size=30000)
tokenizer.train(files=["data.txt"], trainer=trainer)

# Mengkodekan teks sampel
encoded = tokenizer.encode("Hello, how are you?")
print(encoded.tokens)
```

# **Unit 7: Main NLP Tasks**

# **Tujuan:**

Mengenal tugas-tugas utama dalam pemrosesan bahasa alami (NLP) menggunakan model dan pipeline dari Hugging Face.

# **Penjelasan:**

Dalam unit ini, kita akan menggunakan pipeline untuk menyelesaikan beberapa tugas utama NLP:

Analisis Sentimen: Menganalisis apakah sebuah teks memiliki sentimen positif atau negatif.

Menjawab Pertanyaan: Menjawab pertanyaan berdasarkan konteks yang diberikan.

Generasi Teks: Menghasilkan teks baru berdasarkan teks awal yang diberikan.

```python
from transformers import pipeline

# Analisis Sentimen
sentiment_analysis = pipeline("sentiment-analysis")
result = sentiment_analysis("I love Hugging Face!")
print(result)

# Menjawab Pertanyaan
qa_pipeline = pipeline("question-answering")
result = qa_pipeline({
    "question": "What's the name of the company behind Transformers?",
    "context": "Hugging Face is a company based in New York."
})
print(result)

# Generasi Teks
text_generator = pipeline("text-generation")
result = text_generator("Once upon a time, there was a ", max_length=50)
print(result)
```
